# Lesson 2 — Conditionals and Comparisons

Learn how to make decisions in Python using comparisons, `if` / `elif` / `else`, logical operators, and truthiness. Each concept has a short example you can run and modify.

## Comparison operators
`==`, `!=`, `<`, `>`, `<=`, `>=`

```python
print('5 == 5 ->', 5 == 5)
print('5 != 3 ->', 5 != 3)
print('3 < 7  ->', 3 < 7)
print('10 >= 10 ->', 10 >= 10)
```

## if / elif / else
Run different code depending on conditions.

```python
age = 18
if age < 13:
    print('Child')
elif age < 20:
    print('Teenager')
else:
    print('Adult')
```

## Truthiness and logical operators
Values like `0`, `''`, `[]`, and `None` are falsy; most others are truthy. Use `and`, `or`, `not` to combine conditions.

```python
for v in [0, 1, '', 'hi', [], [1], None]:
    print(v, '->', bool(v))

a, b = 5, 10
print('a < b and b < 20 ->', a < b and b < 20)
print('not (a > b) ->', not (a > b))
```

## Short (ternary) conditional and example function
Use a compact form to choose one of two values, and prefer guard clauses inside functions for clarity.

```python
status = 'adult' if age >= 18 else 'minor'
print('Status:', status)

def classify_number(n):
    if n == 0:
        return 'zero'
    if n % 2 == 0:
        return 'even'
    return 'odd'

print(classify_number(0))
print(classify_number(7))
print(classify_number(8))
```
